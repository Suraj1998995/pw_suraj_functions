6. Explain the concept of generators in Python and how they are defined.
Ans:
Generators in Python are special functions that yield values one at a time instead of returning them all at once. They are useful for handling large data sets or streaming data, as they don't store the entire sequence in memory.

How they are defined:
Use the def keyword like a regular function.

Use yield instead of return to produce a value.
Example:

def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for number in count_up_to(3):
    print(number)
-----


7. What are the advantages of using generators over regular functions?
Ans:
Generators offer several advantages over regular functions, especially when dealing with large data or streams:

Memory efficiency:
They yield items one at a time and don’t store the entire output in memory.

Faster startup:
Since they don’t compute all values at once, they can start producing results immediately.

State retention:
They remember where they left off between yields, making them ideal for iterating over large or infinite sequences.

Cleaner code:
Generators can simplify code that would otherwise require complex state tracking with loops and variables.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
Ans:
Here's a simple breakdown of the differences between map(), reduce(), and filter() in Python:

1. map()
Purpose: Applies a function to each item in an iterable.
Returns: A new iterable with the transformed items.

Example:
nums = [1, 2, 3]
result = list(map(lambda x: x * 2, nums))  # [2, 4, 6]

2. filter()
Purpose: Filters items from an iterable based on a condition.
Returns: A new iterable with items that return True from the function.

Example:
nums = [1, 2, 3, 4]
result = list(filter(lambda x: x % 2 == 0, nums))  # [2, 4]

3. reduce() (requires functools module)
Purpose: Applies a function cumulatively to reduce the iterable to a single value.
Returns: A single result.

Example:
from functools import reduce
nums = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, nums)  # 10

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

```python
def sum_of_even_numbers(numbers):
    even_sum = 0
    for num in numbers:
        if num % 2 == 0:
            even_sum += num
    return even_sum

# Example usage
nums = [1, 2, 3, 4, 5, 6]
print(sum_of_even_numbers(nums))  # Output: 12
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

# Example usage
for number in range(1, 201):
    if is_prime(number):
        print(number, "is prime")
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class FibonacciIterator:
    def __init__(self, terms):
        self.terms = terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.terms:
            fib_num = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return fib_num
        else:
            raise StopIteration

# Example usage
fib = FibonacciIterator(10)  # Generate first 10 terms of the Fibonacci sequence
for num in fib:
    print(num)
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

# Example usage
for power in powers_of_two(5):  # Generate powers of 2 up to 2^5
    print(power)
```

9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
# List of temperatures in Celsius
celsius_temps = [0, 20, 30, 40, 100]

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Using map() to apply the conversion function to each element in the list
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

# Output the list of temperatures in Fahrenheit
print(fahrenheit_temps)
```
